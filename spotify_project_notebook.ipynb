```python
from pyspark.sql.functions import col
df = spark.table("main.pmi.spotify_clean")
df.show(5)
```

```python
feature_cols = ["energy", "danceability", "speechiness", "loudness", "tempo", "valence"]
df_ml = df.select(
    *[col(c).cast("double").alias(c) for c in feature_cols],
    col("popularity").cast("double").alias("popularity")
)
df_ml = df_ml.na.drop()
df_ml.show(5)
```

```python
from pyspark.ml.feature import VectorAssembler
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
data = assembler.transform(df_ml).select("features", "popularity")

data.show(5)
```

```python
train, test = data.randomSplit([0.8, 0.2], seed=42)

print("Training_rows:", train.count())
print("Test_rows:", test.count())
```

```python
from pyspark.ml.regression import LinearRegression

lr = LinearRegression(featuresCol="features", labelCol="popularity")

model = lr.fit(train)

print("Model successfully trained!")
print("Coefficients:", model.coefficients)
print("Intercept:", model.intercept)
```

```python
predictions = model.transform(test)

predictions.select("popularity", "prediction").show(10)
```

```python
summary = model.summary

print("RMSE (Root Mean Squared Error):", summary.rootMeanSquaredError)
print("R2 (Coefficient of determination):", summary.r2)
```
